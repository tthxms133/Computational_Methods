# First Derivative






















## Forward Difference

For standard calculations or when only forward points are available:
$$
    f'(x) \approx \boxed{\frac{f(x + h) - f(x)}{h}}.
$$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.sqrt(x)

def F(x):
    return 1 / (2 * np.sqrt(x))

L = 2
N = 5
x = np.linspace(1, L, N)
h = (L - 1) / (N - 1)

f_prime = []
for i in range(N - 1):
    f_prime.append((f(x[i] + h) - f(x[i])) / h)

# Plotting
fig, ax = plt.subplots(figsize=(3, 2))

ax.plot(x, f(x), label=rf'f(x)')
ax.plot(x[:-1], f_prime, label=rf"approximated f'(x)") # Remove last point for plotting
ax.plot(x, F(x), label=rf"true f'(x)")

ax.legend(frameon=False)

plt.tight_layout()
plt.show()
```

## Backward Difference

For bounded domains or near discontinuities:
$$
    f'(x) \approx \boxed{\frac{f(x) - f(x - h)}{h}}.
$$


## Central Difference

Half-steps symmetric around $x$:
$$
    f'(x) \approx \boxed{\frac{f(x + h/2) - f(x - h/2)}{h}}.
$$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.sqrt(x)

def F(x):
    return 1 / (2 * np.sqrt(x))

L = 10
N = 100
x = np.linspace(1, L, N)
h = (L - 1) / (N - 1)

f_prime = []
for i in range(1, N-1): # Only interior points for central difference
    f_prime.append((f(x[i] + (h/2)) - f(x[i] - (h/2))) / h)

# Plotting
fig, ax = plt.subplots(figsize=(3, 2))

ax.plot(x, f(x), label=rf'f(x)')
ax.plot(x[1:-1], f_prime, label=rf"approximated f'(x)", ls=':')
ax.plot(x, F(x), label=rf"true f'(x)", alpha=0.5)

ax.legend(frameon=False)

plt.tight_layout()
plt.show()
```

## Central Difference (Sampled Data)

Data is pre-sampled at intervals of $h$ and half-steps are unavailable:
$$
    f'(x) \approx \boxed{\frac{f(x + h) - f(x - h)}{2h}}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.sqrt(x)

def F(x):
    return 1 / (2 * np.sqrt(x))

L = 10
N = 10
x = np.linspace(1, L, N)
h = (L - 1) / (N - 1)

f_prime = []
for i in range(1, N-1): # Only interior points for central difference
    f_prime.append((f(x[i] + h) - f(x[i] - h)) / (2*h))

# Plotting
fig, ax = plt.subplots(figsize=(3, 2))

ax.plot(x, f(x), label=rf'f(x)')
ax.plot(x[1:-1], f_prime, label=rf"approximated f'(x)", ls=':')
ax.plot(x, F(x), label=rf"true f'(x)", alpha=0.5)

ax.legend(frameon=False)

plt.tight_layout()
plt.show()
```

## Forward/Backward Error

$$
    \epsilon = \boxed{\frac{2C |f(x)|}{h} + \frac{1}{2} h |f''(x)|} \quad \text{and} \quad h_{\text{optimal}} = \boxed{\sqrt{4C \left| \frac{f(x)}{f''(x)} \right|}}.
$$

## Central Error

$$
    \epsilon = \boxed{\frac{2C |f(x)|}{h} + \frac{1}{24} h^2 |f'''(x)| \quad \text{and}} \quad h_{\text{optimal}} = \boxed{\left(24C \left| \frac{f(x)}{f'''(x)} \right|\right)^{1/3}}.
$$

# Second Derivative

$$
    f''(x) \approx \boxed{\frac{f(x + h) - 2f(x) + f(x - h)}{h^2}}.
$$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x ** 5

def F(x):
    return 20 * (x ** 3)

L = 10
N = 100
x = np.linspace(1, L, N)
h = (L - 1) / (N - 1)

f_second = []
for i in range(1, N-1): # Only interior points for central difference
    f_second.append((f(x[i] + h) - 2 * f(x[i]) + f(x[i] - h)) / (h ** 2))

# Plotting
fig, ax = plt.subplots(figsize=(3, 2))

ax.plot(x, f(x), label=rf'f(x)')
ax.plot(x[1:-1], f_second, label=rf"approximated f''(x)", ls=':')
ax.plot(x, F(x), label=rf"true f''(x)", alpha=0.5)

ax.legend(frameon=False)

plt.tight_layout()
plt.show()
```

## Error

$$
    \epsilon = \boxed{\frac{4C |f(x)|}{h} + \frac{1}{12} h^2 |f'''(x)|}, \quad h_{\text{opt}} = \boxed{\left(48C \left| \frac{f(x)}{f'''(x)} \right|\right)^{1/4}}, \quad  \epsilon_{\text{min}} = \boxed{\left(\frac{4}{3} C |f(x) f'''(x)|\right)^{1/2}}.
$$

# Partial Derivative (Central Difference)

$$ 
\begin{align*}
    \frac{\partial f}{\partial x} &\approx \boxed{\frac{f(x + h/2, y) - f(x - h/2, y)}{h}} \\
    \frac{\partial f}{\partial y} &\approx \boxed{\frac{f(x, y + h/2) - f(x, y - h/2)}{h}} \\
    \frac{\partial^2 f}{\partial x \partial y} &\approx \boxed{\frac{f(x + h/2, y + h/2) - f(x - h/2, y + h/2) - f(x + h/2, y - h/2) + f(x - h/2, y - h/2)}{h^2}}.
\end{align*}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return x ** 4 + y - x * y 

def F(x, y):
    return 4 * (x ** 3) - y

def G(x, y):
    return 1 - x

Lx = 10
Nx = 100
x = np.linspace(1, Lx, Nx)
dx = (Lx - 1) / (Nx - 1)

Ly = 100
Ny = 20
y = np.linspace(1, Ly, Ny)
dy = (Ly - 1) / (Ny - 1)

x_c = 5
y_c = 5

fx = []
for i in range(1, Nx-1):
    fx.append((f(x[i] + (dx/2), y_c) - f(x[i] - (dx/2), y_c))/dx)

fy = []
for i in range(1, Ny-1):
    fy.append((f(x_c, y[i] + (dy/2)) - f(x_c, y[i] - (dy/2)))/dy)

# Plotting
fig, ax = plt.subplots(1, 2, figsize=(5, 2))

ax[0].plot(x[1:-1], fx, label=rf"approximated $f_x$", ls=':')
ax[0].plot(x, F(x, y_c), label=rf"true $f_x$", alpha=0.5)
ax[0].set_xlabel(r'$x$')

ax[1].plot(y[1:-1], fy, label=rf"approximated $f_y$", ls=':')
ax[1].plot(y, np.full_like(y, G(x_c, y_c)), label=rf"true $f_y$", alpha=0.5)
ax[1].set_xlabel('$y$')

for i in range(2):
    ax[i].legend(frameon=False)
    ax[i].grid(True, ls=':', alpha=0.3)

plt.tight_layout()
plt.show()
```

# Interpolation




Estimate the value of $f$ at an unknown point $x$ between $a$ and $b$. 

## Linear Interpolation


$$
    f(x) \approx \boxed{\frac{(b - x) f(a) + (x - a) f(b)}{b - a}}.
$$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 3 * x + 2

def g(x):
    return 3 * (x**2) + 2

def y(x, a, b, f):
    return ((b - x) * f(a) + (x - a) * f(b)) / (b - a)

a, b = 3, 4
x = np.linspace(0, 10, 100)

plt.figure(figsize=(3, 2))
plt.plot(x, f(x), ls=':', alpha=0.7, color='r', lw=5)
plt.plot(x, g(x), ls=':', alpha=0.7, color='k', lw=5)
plt.plot(x, y(x, a, b, f), lw=1, color='g', alpha=0.5)
plt.plot(x, y(x, a, b, g), lw=1, color='b', alpha=0.5)

plt.tight_layout()
plt.show()
```

### Approximation Error




$$
    \boxed{\frac{1}{2} (a - x)(b - x) f''(x)}.
$$

### Maximum Error



$$
    \boxed{\frac{1}{8} h^2 |f''(x)|}
$$
